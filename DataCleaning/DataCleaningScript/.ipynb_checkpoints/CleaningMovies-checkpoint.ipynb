```python
import pandas as pd
import numpy as np
path = r"C:\Users\ashle\OneDrive\Desktop\UManitoba\Courses\FALL 2026\COMP 3380\GroupProject\DataCleaning\movies.csv"
```

```python
bad_lines = []

def catch_bad_line(line):
    bad_lines.append(line)
    return None

df = pd.read_csv(path, engine='python', on_bad_lines=catch_bad_line)
```

```python
print(f"Loaded {len(df)} valid rows.")
print(f"Captured {len(bad_lines)} malformed rows for cleaning.")
# Print out the data
df
```

```python
# Data overview
df.info()
print("Data types seems to be consistent and there are no null values.")

print("\nDuplicate rows count:\n", df.duplicated().sum())

print("\nNaN/Null values across all columns:\n", df.isna().sum())
```

```python
# 1. Fix incorrect data type: id shouldn't be a string
bad_id_rows = df[pd.to_numeric(df['id'], errors='coerce').isna()]
bad_id_rows[['id']]

## It seems there's a csv unescaped comma issue so the entire row shifted over.
## Notice that the previous row seems to be the where the broken row should have continued
## Imma save that row by combining them

### Append the split sentence to overview
df.loc[6130, 'overview'] = str(df.loc[6130, 'overview']) + str(df.loc[6131, 'id'])

### Map only the Movie entity attributes from row 6131
df.loc[6130, "release_date"] = df.loc[6131, "title"]
df.loc[6130, "runtime"] = float(df.loc[6131, "original_title"])
df.loc[6130, "budget"] = float(df.loc[6131, "overview"])
df.loc[6130, "status"] = df.loc[6131, "popularity"]  # 'Released'
df.loc[6130, "tagline"] = np.nan
df.loc[6130, "poster_path"] = df.loc[6131, "vote_average"]
df.loc[6130, "backdrop_path"] = df.loc[6131, "vote_count"]

### Drop row 6131 and reset the index
df = df.drop(index=6131).reset_index(drop=True)

### Cast primary key to int
df["id"] = df["id"].astype(int)
```

```python
### Check it's fixed:
df.loc[6130]
```

```python
# 2. Some columns are decided to be grouped as its own entity

## Separate backdrop_path, poster_path, and homepage to MovieAsset entity
assets = []

### Poster paths
posters = df[['id', 'poster_path']].dropna().rename(columns={'poster_path': 'url'})
posters['asset_type'] = 'poster'

### Backdrop paths
backdrops = df[['id', 'backdrop_path']].dropna().rename(columns={'backdrop_path': 'url'})
backdrops['asset_type'] = 'backdrop'

### Homepages
homepages = df[['id', 'homepage']].dropna().rename(columns={'homepage': 'url'})
homepages['asset_type'] = 'homepage'

### Combine and format for MovieAsset(movie_id, asset_type, url)
df_movie_assets = pd.concat([posters, backdrops, homepages], ignore_index=True)
df_movie_assets = df_movie_assets.drop_duplicates()
df_movie_assets = df_movie_assets.rename(columns={'id': 'movieID'})[['movieID', 'url']]

### Save as its own table
df_movie_assets.to_csv("../movie_asset.csv", index=False)
```

```python
# 3. Extract of_genre from Movie table


## Load genres mapping
path = r"C:\Users\ashle\OneDrive\Desktop\UManitoba\Courses\FALL 2026\COMP 3380\GroupProject\DataCleaning\genres.csv"
genres_lookup = pd.read_csv(path).set_index('name')['id'].to_dict()

## Explode the comma-separated genres column to 1 genre per row
movie_genres = df[['id', 'genres']].dropna().copy()
movie_genres['genreName'] = movie_genres['genres'].str.split(',')
movie_genres = movie_genres.explode('genreName')
movie_genres['genreName'] = movie_genres['genreName'].str.strip()

## Map name to genre_id and rename to match ER diagram
movie_genres['genre_id'] = movie_genres['genreName'].map(genres_lookup)
movie_genres = movie_genres.dropna(subset=['genre_id'])
movie_genres['genre_id'] = movie_genres['genre_id'].astype(int)
movie_genres = movie_genres.rename(columns={'id': 'movieID'})[['movieID', 'genre_id']]

## Save the bridge table (of_genre)
movie_genres.to_csv("../of_genre.csv", index=False)
```

```python
# 4. Drop unneeded columns based on ER diagram

## TODO: in our ER diagram, revenue is not included. Why did we not want to include this again? this looks pretty nice to have

## Note: 'backdrop_path', 'poster_path', 'homepage' has been separated to its own entity as above. 'genres' will be its own cleaned entity based on genres.csv

cols_to_drop = [
    'popularity', 'vote_average', 'vote_count',
    'revenue', 'original_language',
    'created_at', 'updated_at', 'video', 'adult',
    'genres', 'backdrop_path', 'poster_path', 'homepage'
]

df = df.drop(columns=[col for col in cols_to_drop if col in df.columns])
df.info()
```

```python
# 5. Fix data formatting to standard YYYY-MM-DD

## Check for non-null values that fail datetime parsing (e.g., corrupted strings/numbers)
parsed_dates = pd.to_datetime(df['release_date'], errors='coerce')
invalid_dates = df[parsed_dates.isna() & df['release_date'].notna()]
print(f"Malformed date strings count: {len(invalid_dates)}")
display(invalid_dates[['id', 'title', 'release_date']])

## Check for missing/null dates
missing_dates = df[df['release_date'].isna()]
print(f"Missing (NaN) date count: {len(missing_dates)}")
display(missing_dates[['id', 'title', 'release_date', 'status']].head())
```

```python
## There's no malformed date but we standardize valid dates to YYYY-MM-DD anyway
df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce').dt.strftime('%Y-%m-%d')

## Replace missing strings with None so pandas exports them as empty SQL NULLs
df['release_date'] = df['release_date'].fillna('None Specified')

print("\nSample Missing Dates:")
df[df['release_date'] == 'None Specified'][['id', 'title', 'release_date', 'status']].head()
```

```python
## Summary count
total_rows = len(df)
none_specified_count = (df['release_date'] == 'None Specified').sum()
valid_date_count = total_rows - none_specified_count

print(f"\nTotal rows: {total_rows}")
print(f"Valid formatted dates: {valid_date_count}")
print(f"Missing ('None Specified') dates: {none_specified_count}")
```

```python
# 6. Check for any zeros in some numeric columns (since they wont' be detected as NaN)

zero_metrics = {
    'zero_runtime': (df['runtime'] == 0).sum(),
    'zero_budget': (df['budget'] == 0).sum(),
}
pd.Series(zero_metrics)
```

```python
## For now set as Null, to prevent producing skewed calculations
df['budget'] = df['budget'].replace(0, np.nan)
df['runtime'] = df['runtime'].replace(0, np.nan)

print(f"Movies with missing budget (set to NULL): {df['budget'].isna().sum()}")
print(f"Movies with 0 runtime remaining: {(df['runtime'] == 0).sum()}")
## Print a few code
df[['title', 'runtime', 'budget', 'status']].sample(5)
```

```python
# 7. Check title for multiple consecutive spaces or leading/trailing whitespace
bad_spaces = df[df['title'].astype(str).str.contains(r'^\s|\s$|\s{2,}', regex=True, na=False)]
bad_spaces[['id', 'title']]
```

```python
## Strip outer spaces and collapse consecutive internal spaces into one
df['title'] = df['title'].astype(str).str.strip().str.replace(r'\s+', ' ', regex=True)
df['original_title'] = df['original_title'].astype(str).str.strip().str.replace(r'\s+', ' ', regex=True)
```

```python
## Re-check one mor
bad_spaces = df[df['title'].astype(str).str.contains(r'^\s|\s$|\s{2,}', regex=True, na=False)]
bad_spaces[['id', 'title']]
```

```python
# 8. Check for any NaNs

df.isna().sum()
```

```python
## Handle missing / default values
df['tagline'] = df['tagline'].fillna('No tagline available')
df['overview'] = df['overview'].fillna('No overview available')

## Check one more time
df.isna().sum()
```

```python
# 9. Final cleanup: renaming

df = df.rename(columns={'id': 'movieID'})

## Reorder columns to align neatly with the ER diagram
movie_cols = [
    'movieID', 'title', 'original_title', 'release_date',
    'runtime', 'budget', 'status', 'tagline', 'overview'
]
df = df[movie_cols]
```

```python
# Save cleaned data in csv
clean_path = r"C:\Users\ashle\OneDrive\Desktop\UManitoba\Courses\FALL 2026\COMP 3380\GroupProject\DataCleaning\CleanDatas\movies_cleaned.csv"
df.to_csv(clean_path, index=False)
```
