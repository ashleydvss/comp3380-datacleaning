```python
bad_lines = []

def catch_bad_line(line):
    bad_lines.append(line)
    return None

df = pd.read_csv(path, engine='python', on_bad_lines=catch_bad_line)
```

```python
print(f"Loaded {len(df)} valid rows.")
print(f"Captured {len(bad_lines)} malformed rows for cleaning.")
# Print out the data
df
```

```python
# Data overview
df.info()
print("Data types seems to be consistent and there are no null values.")

print("Duplicate rows count:", df.duplicated().sum())
```

```python
# Data cleaning start
# 1. TODO: NaN values

print("Sum of NaN values:")
sumOfNaNValues = df.isna().sum()
sumOfNaNValues
```

```python
print("From here, there are 7222 NaN values from Character")
df['character'] = df['character'].fillna('Unknown') #Fill the NaN values to 'unknown' for now.
```

```python
# 2. Drop columns that we don't need from ER diagram

# In the ER diagram, cast (Person) entity only has 4 attributes: personID, personName, castOrder, and character.
# Since we'll need a connection to movie, movie_id would be enough
df = df.drop(columns=['movie_title','movie_release_date'])
df
```

```python
# 3. String format: Trailing whitespace
# Identify rows with two or more consecutive spaces
df_inconsistentWhitespace = df[df['name'].str.contains(r'\s{2,}', regex=True, na=False)]
df_inconsistentWhitespace[['id', 'name']]
```

```python
# Strip leading/trailing spaces, then collapse multiple internal spaces into a single space
df['name'] = df['name'].str.strip().str.replace(r'\s+', ' ', regex=True)

# Confirm leading whitespace has been cleaned
df_inconsistentWhitespace = df[df['name'].str.contains(r'\s{2,}', regex=True, na=False)]
df_inconsistentWhitespace[['id', 'name']]
```
