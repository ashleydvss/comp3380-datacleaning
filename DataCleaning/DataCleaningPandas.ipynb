```python
import pandas as pd
```

```python
path = r"C:\Users\ashle\OneDrive\Desktop\UManitoba\Courses\FALL 2026\COMP 3380\GroupProject\DataCleaning\movies.csv"
```

```python
bad_lines = []

def catch_bad_line(line):
    bad_lines.append(line)
    return None
```

```python
df = pd.read_csv(path, engine='python', on_bad_lines=catch_bad_line)
```

```python
print(f"Loaded {len(df)} valid rows.")
print(f"Captured {len(bad_lines)} malformed rows for cleaning.")
```

```python
# Print out the table
df
```

```python
# Technical jargon: deduplication
df = df.drop_duplicates(subset=['id'], keep='first')
df
```

```python
# Technical jargon: imputation (filling values) and null handling
# Drop rows where critical fields are missing
df = df.dropna(subset=['title', 'release_date'])

# Impute non-critical text fields
df['overview'] = df['overview'].fillna('No overview available')
df['tagline'] = df['tagline'].fillna('No tagline available')
df['homepage'] = df['homepage'].fillna('No homepage available')
df['genres'] = df['genres'].fillna('No genres available')
df
```
